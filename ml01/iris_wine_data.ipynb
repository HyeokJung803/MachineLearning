```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import confusion_matrix, adjusted_rand_score

# 1. 데이터 로드 (4개 특성: 꽃받침 길이/너비, 꽃잎 길이/너비)
iris = load_iris()
X = iris.data
y = iris.target

# 2. K-Means 모델 생성 및 학습 (품종이 3개이므로 clusters=3)
km = KMeans(n_clusters=3, random_state=42)
km.fit(X)

# 3. 혼동 행렬(Confusion Matrix) 출력
print("=== Confusion Matrix ===")
print(confusion_matrix(y, km.labels_))
print()

# 4. Adjusted Rand Score(ARS) 계산 및 출력
ars = adjusted_rand_score(y, km.labels_)
print(f"Adjusted Rand Score (ARS): {ars:.4f}")

# 주석: setosa가 완벽 분리된 이유
# Setosa 품종은 나머지 두 품종(versicolor, virginica)과 비교했을 때 꽃잎과 꽃받침의 크기 특성이 공간상에서 완전히 독립적으로 멀리 떨어져 있기(공간적 고립) 때문이다.
```

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. 데이터 로드 및 데이터프레임 변환
wine = load_wine()
X = wine.data
df = pd.DataFrame(X, columns=wine.feature_names)

# 2. StandardScaler로 스케일링 후 KMeans 학습
# [주석] K-Means는 데이터 간의 거리를 기반으로 군집화를 수행하므로, 각 특성(Feature)의 스케일(단위) 차이가 거리 계산에 왜곡을 주지 않도록 표준화(스케일링)가 필수적이다.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 최적의 군집 수(예: 3)로 K-Means 학습
km = KMeans(n_clusters=3, random_state=42)
km.fit(X_scaled)

# 3. 각 데이터가 속한 군집 중심과의 최소 거리 계산
# km.transform은 각 데이터로부터 모든 군집 중심까지의 거리를 반환하므로, min(axis=1)을 통해 가장 가까운 중심과의 거리를 구합니다.
distances = km.transform(X_scaled).min(axis=1)

# 4. np.percentile로 상위 5% 거리를 가진 데이터를 이상치 임계값으로 설정
threshold = np.percentile(distances, 95)
anomaly_idx = np.where(distances > threshold)[0]

# 5. 탐지된 이상치 특성 값을 df.iloc[anomaly_idx]로 출력
print(f"=== 탐지된 이상치 개수: {len(anomaly_idx)}개 (임계값: {threshold:.4f}) ===")
print(df.iloc[anomaly_idx])
```

1. 스케일링: 모든 특성의 범위를 공평하게
2. K값 지정: km = KMeans(n_clusters=3, random_state=42) 데이터를 몇 개의 그룹으로 나뉠 건지
3. 컴퓨터가 내부적으로 K-Means 방식을 사용해, 서로 최대한 멀리 떨어진 데이터 K개를 무작위로 골라 중심으로 임명.
4. 데이터들은 중심으로 소속, 모인 데이터들로 평균값을 내어 평균을 중심으로
5. 더 이상 결과에 변화가 없을 때까지 수렴
